```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("clean_dataset.csv")
```

```python
#Product Count by Category

category_counts=df["Category"].value_counts()
# Create bar chart
plt.figure(figsize=(8, 5))
bars = plt.bar(category_counts.index, category_counts.values)
plt.bar_label(bars)
plt.title("Product Count by Category")
plt.xlabel("Category")
plt.ylabel("Number of Products")

plt.show()
```

```python
# Calculate average price for each category
avg_price = df.groupby("Category")["Prices"].mean()

# Create bar chart
plt.figure(figsize=(8, 5))
avg_price = plt.bar(avg_price.index, avg_price.values)
plt.bar_label(avg_price)
plt.title("Average Price by Category")
plt.xlabel("Category")
plt.ylabel("Average Price (₹)")

plt.show()
```

```python
# Calculate average discount for each category
avg_discount = df.groupby("Category")["Discount"].mean()

# Create bar chart
plt.figure(figsize=(8, 5))
avg_discount=plt.bar(avg_discount.index, avg_discount.values)
plt.bar_label(avg_discount)
plt.title("Average Discount by Category")
plt.xlabel("Category")
plt.ylabel("Average Discount (%)")

plt.show()
```

```python
#Price vs Discount
#Create scatter chart
plt.figure(figsize=(8, 5))

plt.scatter(df["Prices"], df["Discount"])

plt.title("Price vs Discount")
plt.xlabel("Price (₹)")
plt.ylabel("Discount (%)")

plt.show()
```

```python
# Select top 5 expensive products
top_products = df.nlargest(5, "Prices")

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(top_products["Product_Name"], top_products["Prices"])

plt.title("Top 5 Most Expensive Products")
plt.xlabel("Product")
plt.ylabel("Price (₹)")
plt.xticks(rotation=45)

plt.show()
```

## Task 3: Data Visualization — Conclusion

I created five visualizations to analyze Reliance Digital product data.

1. **Product Count:** Mobile has the highest number of products, while Refrigerators have the lowest.
2. **Average Price:** TVs have the highest average price, while Watches have the lowest.
3. **Average Discount:** TVs offer the highest average discount, while Tablets have the lowest.
4. **Price vs Discount:** Most products are concentrated in the lower price range. The scatter plot does not show a strong linear relationship between price and discount.
5. **Top 5 Expensive Products:** The chart compares the prices of the most expensive products in the dataset.

These visualizations help understand product distribution, pricing patterns, and discounts across different categories.
