```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

data = []

for i in range(1, 75):

    url = f"https://www.reliancedigital.in/collection/mobiles?internal_source=navigation&page_no={i}&page_size=12&page_type=number&sort_on=price_asc&q=mobiles%20under%2050000"

    r = requests.get(url)

    soup = BeautifulSoup(r.text, "html.parser")

    #print("Status:", r.status_code)

    box = soup.find("div", class_="main-container")

    if box is None:
        print("Box not found on page:", i)
        continue

    # Product cards
    products = box.select("div.product-card")

    #print("Page:", i, "Products:", len(products))

    for product in products:

        # Product Name
        name = product.find("div", class_="product-card-title")
        name = name.get_text(strip=True) if name else None

        # Price
        price = product.find("div", class_="price")
        price = price.get_text(strip=True) if price else None

        #Discount
        discount = product.find("div", class_="discount")
        discount = discount.get_text(strip=True) if discount else None


        data.append({
            "Product Name": name,
            "Prices": price,
            "Discount": discount,
            "Category": "Mobile",
            "Date": "2023-09-08",
            "Source": "Reliance Digital"

        })


# Create DataFrame
df = pd.DataFrame(data)

df.to_csv("Reliance_Mobile.csv", index=False)
```

```python
import pandas as pd

# Read all CSV files
mobile = pd.read_csv("Reliance_Mobile.csv")
refrigerator = pd.read_csv("Reliance_Refrigerator.csv")
tv = pd.read_csv("Reliance_TV.csv")
tablets = pd.read_csv("Reliance_Tablets.csv")
watches = pd.read_csv("Reliance_Watches.csv")
laptop = pd.read_csv("Reliance_laptop.csv")

# Combine all datasets row-wise
df = pd.concat(
    [mobile, refrigerator, tv, tablets, watches, laptop],
    ignore_index=True
)

# Save combined dataset
df.to_csv("dataset.csv", index=False, encoding="utf-8-sig")

# Display results
print("All datasets combined successfully!")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print(df.head())
```
