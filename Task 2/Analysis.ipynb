```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Dataset and clean , short and filter dataset
df = pd.read_csv("dataset.csv")
df.shape


#split Product_Name column
parts = df["Product Name"].str.split(r"\s+", n=2, expand=True)

df["Product_Name"] = parts[0] + " " + parts[1].fillna("")
df["Description"] = parts[2]

df["Product_Name"] = df["Product_Name"].str.strip()

#Remove OFF and replace NaN with 0
df["Discount"] = df["Discount"].str.replace("OFF", "", regex=False).str.strip()
df["Discount"] = df["Discount"].str.replace("%", "", regex=False)
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce").fillna(0).astype(int)
df["Discount"] = df["Discount"].fillna("0")

#Price Column Clean
df["Prices"] = df["Prices"].str.replace(r"[^\d.]", "", regex=True)
df["Prices"] = pd.to_numeric(df["Prices"], errors="coerce")

```

```python
# Arrange Columns in Sequence

df = df[[
    "Product Name",
    "Product_Name",
    "Description",
    "Prices",
    "Discount",
    "Category",
    "Date",
    "Source"
]]
```

```python
#Price Analysis
print("Average Price:", df["Prices"].mean()) #Check avg price
print("Minimum Price:", df["Prices"].min())  #Check min Price
print("Maximum Price:", df["Prices"].max())  #check Max price
print("Median Price:", df["Prices"].median()) # check median
```

```python
#Category Analysis
df["Category"].value_counts() # Count products in each category
df.groupby("Category")["Prices"].mean().round(2) # Avg price Category
df.groupby("Category")["Prices"].mean().idxmax() #Highest avg price by category
```

```python
#Discount Analysis

df["Discount"].mean().round(2)  # Average discount
df["Discount"].max()  # Maximum discount
df["Discount"].min()  # Minimum discount
df.groupby("Category")["Discount"].mean().round(2)  # Average discount by category
df.groupby("Category")["Discount"].mean().idxmax()  # Category with highest average discount
df.groupby("Category")["Discount"].max()  # Maximum discount by category
```

```python
# Product Analysis
df.loc[df["Prices"].idxmax()]  # Most expensive product
df.loc[df["Prices"].idxmin()]  # Cheapest product
df.nlargest(5, "Prices")  # Top 5 expensive products
df.nsmallest(5, "Prices")  # Top 5 cheapest products
```

```python
# Product Name Analysis

df["Product_Name"].nunique()  # Total unique products
df["Product_Name"].value_counts().head(10)  # Top 10 frequently appearing products
df["Product_Name"].isnull().sum()  # Missing product names
df["Product_Name"].duplicated().sum()  # Duplicate product names
```

```python
# Price Distribution Analysis

df["Prices"].describe()  # Price distribution summary
df["Prices"].value_counts().head(10)  # Top 10 most frequently occurring prices
df["Prices"].skew().round(2)  # Price distribution skewness
df["Prices"].quantile([0.25, 0.50, 0.75])  # Price quartiles
```

```python
# Relationship Analysis

df[["Prices", "Discount"]].corr().round(2)  # Correlation between price and discount
df.groupby("Category")[["Prices", "Discount"]].mean().round(2)  # Average price and discount by category
df.nlargest(5, "Discount")  # Top 5 products with highest discount
df.nlargest(5, "Prices")[["Product Name", "Prices", "Discount", "Category"]]  # Expensive products with discount
```

```python
# Q1. Which product category has the highest average price?
df.groupby("Category")["Prices"].mean().idxmax()


# Q2. Which product category offers the highest average discount?
df.groupby("Category")["Discount"].mean().idxmax()


# Q3. Which are the top 5 most expensive products?
df.nlargest(5, "Prices")[["Product Name", "Prices", "Category"]]


# Q4. Is there any relationship between product prices and discounts?
df[["Prices", "Discount"]].corr()


# Q5. Which product category contains the highest number of products?
df["Category"].value_counts().idxmax()


# Q6. Are there any unusual or invalid product prices and discount values?
df[(df["Prices"] <= 0) | (df["Discount"] < 0) | (df["Discount"] > 100)]


# Q7. Which products have the highest discounts?
df.nlargest(5, "Discount")[["Product Name", "Discount", "Prices", "Category"]]
```
